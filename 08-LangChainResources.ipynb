# Lang Chain Resources

## Resources

Below are the key resources for learning and working with LangChain, LangGraph, and the broader ecosystem.

---

### 1. LangChain Documentation

The official LangChain documentation — covers installation, core concepts (chains, agents, memory, tools), how-to guides, and integrations with LLM providers.

- **URL:** [python.langchain.com/docs/get_started/introduction](https://python.langchain.com/docs/get_started/introduction)
- Best starting point for understanding the LangChain framework from the ground up.

### 2. LangChain GitHub Repository

The main open-source repository for the LangChain Python library. Browse source code, open issues, contribute, or explore community examples.

- **URL:** [github.com/langchain-ai/langchain](https://github.com/langchain-ai/langchain)

### 3. LangGraph GitHub Repository

LangGraph extends LangChain with stateful, multi-actor graph-based workflows — ideal for building reliable agents and multi-step pipelines.

- **URL:** [github.com/langchain-ai/langgraph](https://github.com/langchain-ai/langgraph)
- This course is built on LangGraph; the repository contains examples, changelogs, and the full API.

### 4. Functions, Tools and Agents with LangChain

A hands-on DeepLearning.AI short course covering OpenAI function calling, LangChain tools and toolkits, and how to build robust agents — a great complement to this LangGraph course.

- **URL:** [deeplearning.ai/short-courses/functions-tools-agents-langchain](https://www.deeplearning.ai/short-courses/functions-tools-agents-langchain/)

### 5. LangChain Hub (Smith)

A community-driven registry of reusable prompts, chains, and agents. Browse, pull, and share prompt templates directly from your code using `langchain hub pull`.

- **URL:** [smith.langchain.com/hub](https://smith.langchain.com/hub)
- Also the home of **LangSmith** — the tracing, evaluation, and monitoring platform for LangChain applications.
